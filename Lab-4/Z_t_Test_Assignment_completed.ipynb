## Quick Reference

| | Z-test | t-test |
|---|---|---|
| Use when | Population σ is **known** | Population σ is **unknown** |
| Formula | $z = \dfrac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}$ | $t = \dfrac{\bar{x} - \mu_0}{s / \sqrt{n}}$, $df = n - 1$ |

| Tail | $H_1$ | p-value (Z) | p-value (t) | Critical value (Z, α = 0.05) |
|---|---|---|---|---|
| Left | $\mu < \mu_0$ | `stats.norm.cdf(z)` | `stats.t.cdf(t, dof)` | −1.645 |
| Right | $\mu > \mu_0$ | `1 - stats.norm.cdf(z)` | `1 - stats.t.cdf(t, dof)` | +1.645 |
| Two | $\mu \neq \mu_0$ | `2 * (1 - stats.norm.cdf(abs(z)))` | `2 * (1 - stats.t.cdf(abs(t), dof))` | ±1.960 |

**Decision rule:** if **p-value ≤ α**, reject $H_0$. Otherwise, fail to reject $H_0$.

For the sample standard deviation use `df['column'].std()`. pandas uses `ddof=1` by default, which is what the t-test needs.

⚠️ The dataset is stored in a variable called `df`. Store the degrees of freedom in a variable called **`dof`**, not `df`, or you will overwrite the dataset.

# Part A – Z-Tests (σ known, use all 40 students)

## Q1. Z-Test – Left-Tailed

A health report says college students sleep **7 hours** per night on average, with a known population standard deviation of **σ = 1.2 hours**.
Using `sleep_hours` for **all 40 students**, test at α = 0.05 whether students sleep **less than 7 hours** on average.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) find $n$ and $\bar{x}$, (c) compute $z$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion.

*(15 marks)*

```python
# ===================== Q1 =====================
mu0 = 7
sigma = 1.2
data = df['sleep_hours']

n = len(data)
xbar = data.mean()
z = (xbar - mu0) / (sigma / np.sqrt(n))

critical_value = -1.645
p_value = stats.norm.cdf(z)

print(f"n = {n}")
print(f"x_bar = {xbar:.4f}")
print(f"z = {z:.4f}")
print(f"critical value = {critical_value}")
print(f"p-value = {p_value:.4f}")
print("Reject H0" if p_value <= alpha else "Fail to reject H0")
```

**Your answer – Q1**

- **$H_0$:** $\mu = 7$ (students sleep 7 hours on average)
- **$H_1$:** $\mu < 7$ (students sleep less than 7 hours on average)
- **Tail (left / right / two):** Left
- **Test statistic:** z = -1.8578
- **Critical value:** -1.645
- **p-value:** 0.0316
- **Decision (Reject $H_0$ / Fail to reject $H_0$):** Reject $H_0$ (p = 0.0316 ≤ α = 0.05, and z = -1.8578 is more extreme than -1.645)
- **Conclusion in words:** There is enough evidence at the 5% significance level to conclude that these 40 students sleep, on average, less than 7 hours per night.

## Q2. Z-Test – Right-Tailed

A national survey says young adults spend **5 hours** a day on screens, with a known population standard deviation of **σ = 1.5 hours**.
Using `screen_time` for **all 40 students**, test at α = 0.05 whether these students' screen time is **more than 5 hours** a day.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) find $n$ and $\bar{x}$, (c) compute $z$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion.

*(15 marks)*

```python
# ===================== Q2 =====================
mu0 = 5
sigma = 1.5
data = df['screen_time']

n = len(data)
xbar = data.mean()
z = (xbar - mu0) / (sigma / np.sqrt(n))

critical_value = 1.645
p_value = 1 - stats.norm.cdf(z)

print(f"n = {n}")
print(f"x_bar = {xbar:.4f}")
print(f"z = {z:.4f}")
print(f"critical value = {critical_value}")
print(f"p-value = {p_value:.4f}")
print("Reject H0" if p_value <= alpha else "Fail to reject H0")
```

**Your answer – Q2**

- **$H_0$:** $\mu = 5$ (screen time is 5 hours a day on average)
- **$H_1$:** $\mu > 5$ (screen time is more than 5 hours a day on average)
- **Tail (left / right / two):** Right
- **Test statistic:** z = 2.0028
- **Critical value:** 1.645
- **p-value:** 0.0226
- **Decision (Reject $H_0$ / Fail to reject $H_0$):** Reject $H_0$ (p = 0.0226 ≤ α = 0.05, and z = 2.0028 exceeds 1.645)
- **Conclusion in words:** There is enough evidence at the 5% significance level that these students' average daily screen time is more than 5 hours.

## Q3. Z-Test – Two-Tailed

Over many years, the course's final exam has had a mean of **65** with a known standard deviation of **σ = 10**.
Using `exam_score` for **all 40 students**, test at α = 0.05 whether this year's mean score is **different from 65**.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) find $n$ and $\bar{x}$, (c) compute $z$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion.

*(15 marks)*

```python
# ===================== Q3 =====================
mu0 = 65
sigma = 10
data = df['exam_score']

n = len(data)
xbar = data.mean()
z = (xbar - mu0) / (sigma / np.sqrt(n))

critical_value = (-1.96, 1.96)
p_value = 2 * (1 - stats.norm.cdf(abs(z)))

print(f"n = {n}")
print(f"x_bar = {xbar:.4f}")
print(f"z = {z:.4f}")
print(f"critical values = {critical_value}")
print(f"p-value = {p_value:.4f}")
print("Reject H0" if p_value <= alpha else "Fail to reject H0")
```

**Your answer – Q3**

- **$H_0$:** $\mu = 65$ (mean exam score equals 65)
- **$H_1$:** $\mu \neq 65$ (mean exam score differs from 65)
- **Tail (left / right / two):** Two
- **Test statistic:** z = -0.5060
- **Critical value:** ±1.960
- **p-value:** 0.6129
- **Decision (Reject $H_0$ / Fail to reject $H_0$):** Fail to reject $H_0$ (p = 0.6129 > α = 0.05, and |z| = 0.5060 does not exceed 1.960)
- **Conclusion in words:** There is not enough evidence at the 5% significance level to say this year's mean exam score is different from the historical mean of 65.

## Q4. t-Test – Left-Tailed

The class teacher of **Section B** believes her students' attendance is **below 83 %**. The population standard deviation is **not known**.
Using `attendance` for **Section B only**, test at α = 0.05 whether the mean attendance is **less than 83 %**.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) select the rows for the section, then find $n$, $\bar{x}$, $s$ and $df$, (c) compute $t$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion, (f) check your answer with `stats.ttest_1samp(data, popmean, alternative=...)`.

*(15 marks)*

```python
# ===================== Q4 =====================
mu0 = 83
data = df[df['section'] == 'B']['attendance']

n = len(data)
xbar = data.mean()
s = data.std()
dof = n - 1
t_stat = (xbar - mu0) / (s / np.sqrt(n))

critical_value = -stats.t.ppf(0.95, dof)
p_value = stats.t.cdf(t_stat, dof)

print(f"n = {n}, x_bar = {xbar:.4f}, s = {s:.4f}, dof = {dof}")
print(f"t = {t_stat:.4f}")
print(f"critical value = {critical_value:.4f}")
print(f"p-value = {p_value:.4f}")
print("Reject H0" if p_value <= alpha else "Fail to reject H0")

# (f) check with scipy
check = stats.ttest_1samp(data, mu0, alternative='less')
print("\nscipy check:", check)
```

**Your answer – Q4**

- **$H_0$:** $\mu = 83$ (Section B mean attendance is 83%)
- **$H_1$:** $\mu < 83$ (Section B mean attendance is below 83%)
- **Tail (left / right / two):** Left
- **Test statistic:** t = -1.0307 (n = 13, $\bar{x}$ = 80.6154, s = 8.3420, df = 12)
- **Critical value:** -1.7823
- **p-value:** 0.1615
- **Decision (Reject $H_0$ / Fail to reject $H_0$):** Fail to reject $H_0$ (p = 0.1615 > α = 0.05, and t = -1.0307 is not more extreme than -1.7823)
- **Conclusion in words:** There is not enough evidence at the 5% significance level that Section B's mean attendance is below 83%. The scipy check (`ttest_1samp`, alternative='less') gives the same t and p-value.

## Q5. t-Test – Right-Tailed

The college recommends at least **2.5 hours** of self-study per day. The teacher of **Section A** claims her students study **more than 2.5 hours** a day on average. σ is **not known**.
Using `study_hours` for **Section A only**, test the claim at α = 0.05.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) select the rows for the section, then find $n$, $\bar{x}$, $s$ and $df$, (c) compute $t$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion, (f) check your answer with `stats.ttest_1samp(data, popmean, alternative=...)`.

*(15 marks)*

```python
# ===================== Q5 =====================
mu0 = 2.5
data = df[df['section'] == 'A']['study_hours']

n = len(data)
xbar = data.mean()
s = data.std()
dof = n - 1
t_stat = (xbar - mu0) / (s / np.sqrt(n))

critical_value = stats.t.ppf(0.95, dof)
p_value = 1 - stats.t.cdf(t_stat, dof)

print(f"n = {n}, x_bar = {xbar:.4f}, s = {s:.4f}, dof = {dof}")
print(f"t = {t_stat:.4f}")
print(f"critical value = {critical_value:.4f}")
print(f"p-value = {p_value:.4f}")
print("Reject H0" if p_value <= alpha else "Fail to reject H0")

# (f) check with scipy
check = stats.ttest_1samp(data, mu0, alternative='greater')
print("\nscipy check:", check)
```

**Your answer – Q5**

- **$H_0$:** $\mu = 2.5$ (Section A mean study hours is 2.5)
- **$H_1$:** $\mu > 2.5$ (Section A mean study hours is more than 2.5)
- **Tail (left / right / two):** Right
- **Test statistic:** t = 2.1185 (n = 14, $\bar{x}$ = 2.9286, s = 0.7569, df = 13)
- **Critical value:** 1.7709
- **p-value:** 0.0270
- **Decision (Reject $H_0$ / Fail to reject $H_0$):** Reject $H_0$ (p = 0.0270 ≤ α = 0.05, and t = 2.1185 exceeds 1.7709)
- **Conclusion in words:** There is enough evidence at the 5% significance level to support the teacher's claim that Section A students study more than 2.5 hours a day on average. The scipy check (`ttest_1samp`, alternative='greater') gives the same t and p-value.

## Q6. t-Test – Two-Tailed

The transport office plans the college bus timetable assuming an average one-way commute of **35 minutes**. σ is **not known**.
Using `commute_min` for **Section C only**, test at α = 0.05 whether the mean commute time is **different from 35 minutes**.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) select the rows for the section, then find $n$, $\bar{x}$, $s$ and $df$, (c) compute $t$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion, (f) check your answer with `stats.ttest_1samp(data, popmean, alternative=...)`.

*(15 marks)*

```python
# ===================== Q6 =====================
mu0 = 35
data = df[df['section'] == 'C']['commute_min']

n = len(data)
xbar = data.mean()
s = data.std()
dof = n - 1
t_stat = (xbar - mu0) / (s / np.sqrt(n))

critical_value = stats.t.ppf(0.975, dof)
p_value = 2 * (1 - stats.t.cdf(abs(t_stat), dof))

print(f"n = {n}, x_bar = {xbar:.4f}, s = {s:.4f}, dof = {dof}")
print(f"t = {t_stat:.4f}")
print(f"critical values = +/-{critical_value:.4f}")
print(f"p-value = {p_value:.4f}")
print("Reject H0" if p_value <= alpha else "Fail to reject H0")

# (f) check with scipy
check = stats.ttest_1samp(data, mu0, alternative='two-sided')
print("\nscipy check:", check)
```

**Your answer – Q6**

- **$H_0$:** $\mu = 35$ (Section C mean commute time is 35 minutes)
- **$H_1$:** $\mu \neq 35$ (Section C mean commute time differs from 35 minutes)
- **Tail (left / right / two):** Two
- **Test statistic:** t = -2.8611 (n = 13, $\bar{x}$ = 29.2308, s = 7.2704, df = 12)
- **Critical value:** ±2.1788
- **p-value:** 0.0143
- **Decision (Reject $H_0$ / Fail to reject $H_0$):** Reject $H_0$ (p = 0.0143 ≤ α = 0.05, and |t| = 2.8611 exceeds 2.1788)
- **Conclusion in words:** There is enough evidence at the 5% significance level that Section C's mean commute time is different from (in fact, shorter than) 35 minutes, so the transport office's 35-minute assumption should be revisited. The scipy check (`ttest_1samp`, alternative='two-sided') gives the same t and p-value.
